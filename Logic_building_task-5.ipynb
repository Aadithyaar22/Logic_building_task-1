```python
def analyze_post_engagement(likes_list):

    total_likes = 0

    for likes in likes_list:
        total_likes += likes

    print("Total Likes:", total_likes)

    if total_likes >= 1000:
        print("Post Status: Viral Post")
    else:
        print("Post Status: Normal Engagement")


# Example
likes = [200, 300, 150, 400]
analyze_post_engagement(likes)
```

```python
def check_medicine_stock(stock):

    print("Medicine Stock:", stock)

    if stock < 10:
        print("Status: Low Stock Alert")
    else:
        print("Status: Stock Sufficient")


# Example
check_medicine_stock(6)
```

```python
def check_rainfall(rainfall_data, required_level):

    total = 0

    for rain in rainfall_data:
        total += rain

    average = total / len(rainfall_data)

    print("Average Rainfall:", average)

    if average >= required_level:
        print("Rainfall Status: Adequate Rainfall")
    else:
        print("Rainfall Status: Inadequate Rainfall")


# Example
rainfall = [70, 75, 68, 75]
check_rainfall(rainfall, 70)
```

```python
def detect_duplicates(usernames):

    if len(usernames) != len(set(usernames)):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")


# Example
accounts = ["user1", "user2", "user3", "user1"]
detect_duplicates(accounts)
```

```python
def filter_premium_crops(prices):

    premium = []

    for price in prices:
        if price > 2000:
            premium.append(price)

    print("Premium Crops:", premium)


# Example
crop_prices = [1500, 2500, 1800, 3200]
filter_premium_crops(crop_prices)
```

```python
def check_system_health(errors):

    print("Error Count:", errors)

    if errors == 0:
        print("System Status: Healthy")
    elif errors <= 5:
        print("System Status: Minor Issues")
    else:
        print("System Status: Critical Issues")


# Example
check_system_health(7)
```

```python
def check_exam_eligibility(attendance_list):

    present = 0

    for day in attendance_list:
        if day == "P":
            present += 1

    percentage = (present / len(attendance_list)) * 100

    print("Attendance Percentage:", percentage)

    if percentage >= 75:
        print("Exam Eligibility: Eligible")
    else:
        print("Exam Eligibility: Not Eligible")


# Example
attendance = ["P", "P", "A", "P", "P"]
check_exam_eligibility(attendance)
```

```python
def calculate_electricity_bill(units):

    bill = 0

    if units <= 100:
        bill = units * 3

    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)

    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)

    # Usage classification
    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"

    return bill, usage


# Example
units = 250
bill_amount, status = calculate_electricity_bill(units)

print("Total Bill:", bill_amount)
print("Usage Classification:", status)
```
