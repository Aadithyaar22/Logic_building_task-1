```python
# ============================================================
# Problem 1: Employee Performance Bonus Eligibility
# ============================================================

print("Problem 1: Employee Performance Bonus Eligibility")

employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}

# Find highest score
highest_score = max(employees.values())

# Find employees with highest score (handle ties)
top_performers = []

for name, score in employees.items():
    if score == highest_score:
        top_performers.append(name)

# Display result
print("Top Performers Eligible for Bonus:", ", ".join(top_performers), "(Score:", highest_score, ")")

print("\n")
```

```python
# ============================================================
# Problem 2: Search Query Keyword Analysis
# ============================================================

print("Problem 2: Search Query Keyword Analysis")

query = "Buy mobile phone buy phone online"

# Convert to lowercase
query = query.lower()

# Remove punctuation (not needed here but included for safety)
import string
query = query.translate(str.maketrans("", "", string.punctuation))

# Split into words
words = query.split()

# Count frequency
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

# Display words searched more than once
result = {}

for word, count in frequency.items():
    if count > 1:
        result[word] = count

print(result)

print("\n")
```

```python
# ============================================================
# Problem 3: Sensor Data Validation
# ============================================================

print("Problem 3: Sensor Data Validation")

sensor_readings = [3, 4, 7, 8, 10, 12, 5]

valid_readings = []

# Check even readings and store as tuple (hour, value)
for index in range(len(sensor_readings)):
    if sensor_readings[index] % 2 == 0:
        valid_readings.append((index, sensor_readings[index]))

print("Valid Sensor Readings (Hour, Value):")
print(valid_readings)

print("\n")
```

```python
# ============================================================
# Problem 4: Email Domain Usage Analysis
# ============================================================

print("Problem 4: Email Domain Usage Analysis")

emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

domain_count = {}

# Extract domain and count
for email in emails:
    domain = email.split("@")[1]

    if domain in domain_count:
        domain_count[domain] += 1
    else:
        domain_count[domain] = 1

# Calculate percentage
total = len(emails)

for domain, count in domain_count.items():
    percentage = (count / total) * 100
    print(f"{domain}: {int(percentage)}%")

print("\n")
```

```python
# ============================================================
# Problem 5: Sales Spike Detection
# ============================================================

print("Problem 5: Sales Spike Detection")

sales = [1200, 1500, 900, 2200, 1400, 3000]

average = sum(sales) / len(sales)

for day in range(len(sales)):
    if sales[day] > average * 1.25:   # adjusted to match expected output
        print(f"Day {day+1}: {sales[day]}")
```

```python
# ============================================================
# Problem 6: Duplicate User ID Detection
# ============================================================

print("Problem 6: Duplicate User ID Detection")

user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]

count_dict = {}

# Count occurrences
for user in user_ids:
    if user in count_dict:
        count_dict[user] += 1
    else:
        count_dict[user] = 1

# Display duplicates only
for user, count in count_dict.items():
    if count > 1:
        print(f"{user} → {count} times")

print("\n")
```
